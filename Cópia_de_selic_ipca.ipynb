```python
import matplotlib.pyplot as plt


anos = [2012, 2013, 2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023]
selic = [7.25, 10.0, 11.0, 14.25, 13.75, 7.0, 6.5, 4.5, 2.0, 9.25, 13.75, 13.0]
ipca = [5.84, 5.91, 6.41, 10.67, 6.29, 2.95, 3.75, 4.31, 4.52, 10.06, 5.79, 4.62]



plt.figure(figsize=(11,6))
plt.grid(True)
plt.plot(anos, selic, label='Selic')
plt.plot(anos, ipca, label='IPCA')
plt.legend()


plt.ylim(0, 16)
plt.xticks([2012, 2014, 2016,2018, 2020, 2022])
plt.plot(2015, 14.25, marker='o',color='blue')
plt.plot(2015, 10.67, marker='o',color='orange')

plt.text(2015, 14.25, '14.25%', color='blue')
plt.text(2015, 10.67, '10.67%', color='orange')

plt.title('Análise IPCA x Selic')
plt.xlabel('Ano')
plt.ylabel('Taxa Anual (%)')




plt.show()
```
